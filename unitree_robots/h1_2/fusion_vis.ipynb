```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('./start_fusion_data.csv')

fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(8, 12))

df.plot(x='t', y='roll', ax=ax1, color='red', title='Roll')
ax1.set_ylabel('roll')
ax1.grid(True, alpha=0.3)
 
df.plot(x='t', y='pitch', ax=ax2, color='green', title='Pitch')
ax2.set_ylabel('pitch')
ax2.grid(True, alpha=0.3)

df.plot(x='t', y='yaw', ax=ax3, color='blue', title='yaw')
ax3.set_ylabel('yaw')
ax3.set_xlabel('Time (t)')
ax3.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
