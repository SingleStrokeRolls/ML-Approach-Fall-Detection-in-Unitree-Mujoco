```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('./csv/0413_chair_falling.csv')
df['acc_norm'] = np.linalg.norm(df[['acc_x', 'acc_y', 'acc_z']], axis=1)
df['gyro_norm'] = np.linalg.norm(df[['gyro_x', 'gyro_y', 'gyro_z']], axis=1)
df.head()
```

```python
target_time_sim = df.loc[df[df['t'].between(1.5, 2)]['acc_norm'].idxmin(), 't']

fig, axs = plt.subplots(2, 2, figsize=(8,6))
df.plot(x='t', y=['acc_x','acc_y','acc_z'], ax=axs[0,0])
df.plot(x='t', y='acc_norm', ax=axs[1,0])
df.plot(x='t', y=['gyro_x','gyro_y','gyro_z'], ax=axs[0,1])
df.plot(x='t', y='gyro_norm', ax=axs[1,1])

axs[0,0].axvline(x=target_time_sim, color='red', linestyle='--', alpha=0.5)
axs[0,1].axvline(x=target_time_sim, color='red', linestyle='--', alpha=0.5)
axs[1,0].axvline(x=target_time_sim, color='red', linestyle='--', alpha=0.5)
axs[1,1].axvline(x=target_time_sim, color='red', linestyle='--', alpha=0.5)
```

```python
target_time_sim = df.loc[df[df['t'].between(2.1, 2.19)]['acc_z'].idxmin(), 't']

df_filtered = df[df['t'] > 1.55]

fig, axs = plt.subplots(2, 2, figsize=(8,6))

df_filtered.plot(x='t', y=['acc_x','acc_y','acc_z'], ax=axs[0,0])
df_filtered.plot(x='t', y='acc_norm', ax=axs[1,0])
df_filtered.plot(x='t', y=['gyro_x','gyro_y','gyro_z'], ax=axs[0,1])
df_filtered.plot(x='t', y='gyro_norm', ax=axs[1,1])

axs[0,0].axvline(x=target_time_sim, color='red', linestyle='--', alpha=0.5)
axs[0,1].axvline(x=target_time_sim, color='red', linestyle='--', alpha=0.5)
axs[1,0].axvline(x=target_time_sim, color='red', linestyle='--', alpha=0.5)
axs[1,1].axvline(x=target_time_sim, color='red', linestyle='--', alpha=0.5)
```
